```python
# ==========================================
# 2. SEPARATE BENIGN AND MALICIOUS
# ==========================================

# --- Dynamic Side ---
# Class 0 = Benign, Class 1 = Malware
dyn_benign = df_dynamic[df_dynamic['Class'] == 0]

# Filter specifically for SPYWARE (Class 1 + Category contains "Spyware")
# We use case=False to match "Spyware", "spyware", etc.
dyn_spyware = df_dynamic[
    (df_dynamic['Class'] == 1) & 
    (df_dynamic['Category'].str.contains("Spyware", case=False, na=False))
]
```

```python
# --- Network Side ---
# Label 0 = Benign, Label 1 = Malicious
net_benign = df_network[df_network[' Label'] == 0]
net_malicious = df_network[df_network[' Label'] == 1]

print(f"\n--- Dynamic Counts ---")
print(f"Benign: {len(dyn_benign)}")
print(f"Spyware: {len(dyn_spyware)}")

print(f"\n--- Network Counts ---")
print(f"Benign: {len(net_benign)}")
print(f"Malicious: {len(net_malicious)}")
```

```python
# ==========================================
# 3. CREATE BALANCED PAIRS
# ==========================================

# Determine how many samples we can create (limited by the smallest dataset)
n_safe = min(len(dyn_benign), len(net_benign))
n_attack = min(len(dyn_spyware), len(net_malicious))

print(f"\nCreating {n_safe} Safe Pairs and {n_attack} Attack Pairs...")

# --- Create Safe Dataset (Benign Process + Benign Network) ---
# Randomly sample and reset index so they align side-by-side
safe_dyn = dyn_benign.sample(n=n_safe, random_state=42).reset_index(drop=True)
safe_net = net_benign.sample(n=n_safe, random_state=42).reset_index(drop=True)

# Concatenate columns (axis=1)
# Note: We drop the 'Label' from network to avoid duplicate label columns, 
# keeping 'Class' as the master label if you want, or keep both.
df_safe = pd.concat([safe_dyn, safe_net], axis=1)

# --- Create Attack Dataset (Spyware Process + Malicious Network) ---
attack_dyn = dyn_spyware.sample(n=n_attack, random_state=42).reset_index(drop=True)
attack_net = net_malicious.sample(n=n_attack, random_state=42).reset_index(drop=True)

df_attack = pd.concat([attack_dyn, attack_net], axis=1)
```

```python
# ==========================================
# 4. MERGE & SAVE
# ==========================================

# Combine Safe and Attack rows (axis=0)
df_final = pd.concat([df_safe, df_attack], axis=0)

# Shuffle the final dataset so safe/attack are mixed
df_final = df_final.sample(frac=1, random_state=42).reset_index(drop=True)
```
