```python
import numpy as np
from sklearn.feature_selection import SelectKBest, f_classif

X = final_df.drop(' Label',axis=1)
y = final_df[' Label']
```

```python

from sklearn.feature_selection import SelectKBest, mutual_info_classif
from sklearn.preprocessing import StandardScaler
# Remove columns where all values are the same
X = X.loc[:, X.nunique() > 1]
print(f"Shape after dropping constant columns: {X.shape}")

# 2. Scale the data (Crucial for Feature Selection and MLP)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Use Mutual Information instead of f_classif
# Why? f_classif only checks for LINEAR relationships. 
# Malware patterns are often NON-LINEAR. Mutual Info captures these better.
# We increase k to 25 to give the Deep Learning model more signal to work with.
k = 25 
selector = SelectKBest(score_func=mutual_info_classif, k=k)
X_new = selector.fit_transform(X_scaled, y)

# 4. Get the clean list of selected features
selected_indices = selector.get_support(indices=True)
selected_features = X.columns[selected_indices]

print("\nTop 25 Selected Features for Stealth Detection:")
print(selected_features.tolist())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
# --- STEP 2: DEFINE THE DEEP LEARNING MODEL ---
# MLPClassifier with hidden layers = Deep Learning
mlp_model = MLPClassifier(
    hidden_layer_sizes=(64, 32), # Layer 1: 64 neurons, Layer 2: 32 neurons
    activation='relu',           # Standard activation for modern DL
    solver='adam',               # Optimizer
    alpha=0.0001,                # Regularization to prevent overfitting
    batch_size=64,               # Process 64 rows at a time
    learning_rate_init=0.001,    
    max_iter=500,                # Max epochs
    early_stopping=True,         # Stop if it stops improving (saves time)
    validation_fraction=0.1,     # Use 10% of train data for validation
    random_state=42,
    verbose=True                 # Show training log
)
```

```python
# --- STEP 1: PREPARE DATA ---
# We use your 'df_new' (25 selected features) and 'y' (labels)
# It is CRITICAL to scale data for Neural Networks
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_new)

# Split into Train (80%) and Test (20%)
# stratify=y ensures we keep the same ratio of Benign/Bot/Infiltration in both sets
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Training Data Shape: {X_train.shape}")
print(f"Testing Data Shape: {X_test.shape}")
```

```python
# --- STEP 2: DEFINE THE DEEP LEARNING MODEL ---
mlp_model = MLPClassifier(
    hidden_layer_sizes=(64, 32), # Layer 1: 64 neurons, Layer 2: 32 neurons
    activation='relu',           # Standard activation for modern DL
    solver='adam',               # Optimizer
    alpha=0.0001,                # Regularization to prevent overfitting
    batch_size=64,               # Process 64 rows at a time
    learning_rate_init=0.001,    
    max_iter=500,                # Max epochs
    early_stopping=True,         # Stop if it stops improving (saves time)
    validation_fraction=0.1,     # Use 10% of train data for validation
    random_state=42,
    verbose=True                 # Show training log
)
```

```python
# --- STEP 3: TRAIN ---
print("\nStarting Training (this might take a minute)...")
mlp_model.fit(X_train, y_train)
print("Training Complete!")
```

```python
# --- STEP 4: EVALUATE ---
y_pred = mlp_model.predict(X_test)
# Calculate Accuracy
acc = accuracy_score(y_test, y_pred)
print(f"\n--- Model Accuracy: {acc*100:.2f}% ---\n")

# Detailed Report
# We map the numeric labels back to names for clarity
target_names = ['Benign', 'Bot', 'Infiltration'] 

print(classification_report(y_test, y_pred, target_names=target_names))
```

```python
# --- STEP 5: VISUALIZE CONFUSION MATRIX ---
plt.figure(figsize=(8, 6))
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=target_names,
            yticklabels=target_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix: Stealthy Malware Detection (MLP)')
plt.show()
```

```python
# --- STEP 6: LOSS CURVE ---
# This proves the model is learning (Loss should go DOWN)
plt.figure(figsize=(8, 5))
plt.plot(mlp_model.loss_curve_)
plt.title('Deep Learning Training Loss')
plt.xlabel('Iterations')
plt.ylabel('Loss')
plt.show()
```

```python
import numpy as np
from sklearn.pipeline import Pipeline
from skl2onnx import to_onnx
from skl2onnx.common.data_types import FloatTensorType

# --- STEP 1: CREATE THE PIPELINE ---
# This bundles your Scaler and Model together so the ONNX file handles scaling automatically.
# We assume 'scaler' and 'mlp_model' are the variables from your training steps.
pipeline = Pipeline([
    ('scaler', scaler),      # The StandardScaler you fitted earlier
    ('mlp', mlp_model)       # The MLPClassifier you trained
])

# --- STEP 2: DEFINE INPUT TYPE ---
# We force the input to be Float32 (compatible with ONNX Runtime)
# n_features should be 25 based on your dataset
n_features = 25 
initial_type = [('float_input', FloatTensorType([None, n_features]))]

# --- STEP 3: CONVERT TO ONNX ---
# We use a sample of data converted to float32 to help the converter understand the shape
print("Converting model to ONNX...")
onx_pipeline = to_onnx(
    pipeline, 
    X_train[:1].astype(np.float32), 
    initial_types=initial_type,
    target_opset=12,
    options={'zipmap': False}  # <--- This fixes the "INVALID_GRAPH" error
)

# --- STEP 4: SAVE THE FILE ---
output_path = "stealthy_malware_pipeline.onnx"
with open(output_path, "wb") as f:
    f.write(onx_pipeline.SerializeToString())

print(f"SUCCESS: Pipeline saved to {output_path}")
print("Move this file to your project folder where 'network_layer.py' is located.")
```
