```python
import pandas as pd
import random
import string


def format_malware_names(category_val):
    """
    Leaves 'Benign' untouched.
    Converts 'Spyware-CWS-hash.raw' -> ('Spyware-A.Tesla-random.exe', 'Spyware-A.Tesla')
    """
    # Handle any blank/null rows safely
    if pd.isna(category_val):
        return pd.Series(['Unknown', 'Unknown'])
        
    cat_str = str(category_val).strip()
    
    # Rule 1: Do not change Benign traffic
    if cat_str == 'Benign':
        return pd.Series(['Benign', 'Benign'])
    
    # Rule 2: Process Malware
    # Split the string by '-' to extract the root class (e.g., 'Spyware')
    parts = cat_str.split('-')
    malware_class = parts[0] if len(parts) > 0 else "Malware"
    
    # Create the clean category (e.g., "Spyware-A.Tesla")
    category_clean = f"{malware_class}-A.Tesla"
    
    # Generate a random 6-character suffix (e.g., "2to1ep") to simulate different executables
    random_suffix = ''.join(random.choices(string.ascii_lowercase + string.digits, k=6))
    
    # Create the full category name (e.g., "Spyware-A.Tesla-2to1ep.exe")
    category_full = f"{category_clean}-{random_suffix}.exe"
    
    return pd.Series([category_full, category_clean])

print("Processing dataset... Please wait.")

# 2. Apply the function to the dataframe
# This updates the 'Category' column and creates/updates the 'Category_clean' column simultaneously
df[['Category', 'Category_clean']] = df['Category'].apply(format_malware_names)

# 3. Save the cleaned dataset to a new file
output_file = 'hybrid_malware_cleaned.csv'
df.to_csv(output_file, index=False)

print(f"Success! Cleaned dataset saved as '{output_file}'")
```
